```python
import re

text11 = re.sub("<br /><br />","", text1)
```

```python
re.sub(r'[^a-zA-Z0-9.\'\s]', '', text11)
```

```python
def preprocess_data(text):
    text1 = re.sub("<br /><br />","", text)
    text2 = re.sub(r'[^a-zA-Z0-9.\'\s]', '', text1)

    return text2
```

```python
preprocess_data(text1)
```

```python
df['Review'] = df['review'].map(lambda x: preprocess_data(x))
```
